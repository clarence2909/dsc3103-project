```python
import matplotlib.pyplot as plt

prices_df = pd.read_csv("../data/raw/prices.csv")
prices_df["price"] = pd.to_numeric(prices_df["price"], errors="coerce")

plt.figure(figsize=(8, 5))
plt.hist(prices_df["price"].dropna(), bins=20)
plt.title("Distribution of Price (raw data)")
plt.xlabel("Price")
plt.ylabel("Number of rows")
plt.grid(True)
plt.savefig("../docs/price_histogram.png")
plt.show()
```
